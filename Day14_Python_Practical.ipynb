Task 3: Create the List Marks=[78,85,90,72,95]. Print the Total,The Average,The Highest and The Lowest

```python
marks = [78, 85, 90, 40, 95]

print("Marks:", marks)
print("Total:", sum(marks))
print("Average:", sum(marks) / len(marks))
print("Highest mark:", max(marks))
print("Lowest mark:", min(marks))
```

Task 4: Write a Function Average(numbers) that returns tha average. Test it with the list in exercise 3.

```python
def average(numbers):
    return sum(numbers) / len(numbers)

print("Average of marks:", average(marks))
```

Task 5: Write a loop that prints each mark with the word Pass if it is 50 or more, otherwise Fail.

```python
for mark in marks:
    if mark >= 50:
        print(mark, "- Pass")
    else:
        print(mark, "- Fail")
```

Task 6: Create the Dictionary of product sales from section 14.4.  Print each product and its units, then the total of all units.

```python
product_sales = {
    "Brownies": 85,
    "Cupcakes": 60,
    "Cakes": 40,
    "Cookies": 25,
    "Pastries": 50
}

print("Product Sales:")

for product, units in product_sales.items():
    print(product, ":", units, "units")

total_units = sum(product_sales.values())

print("Total units sold:", total_units)
```

Task 7: Using the Monthly Sales List, count how many months had sales above 50.

```python
monthly_sales = [45, 62, 55, 38, 71, 50, 84, 49, 66, 30, 58, 92]

months_above_50 = 0

for sales in monthly_sales:
    if sales > 50:
        months_above_50 += 1

print("Monthly sales:", monthly_sales)
print("Number of months with sales above 50:", months_above_50)
```

Task 8: Ask the user for their marks with input() and print the grade (A, B, C or D).

```python
score = float(input("Enter your marks (0 to 100): "))

if score < 0 or score > 100:
    print("Please enter marks between 0 and 100.")
elif score >= 80:
    print("Grade: A")
elif score >= 70:
    print("Grade: B")
elif score >= 50:
    print("Grade: C")
else:
    print("Grade: F")
```

Task 9: Create a pandas DataFrame from the monthly sales and print the total, average, and maximum.

```python
import pandas as pd

months = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

sales_df = pd.DataFrame({
    "Month": months,
    "Sales": monthly_sales
})

display(sales_df)

print("Total sales:", sales_df["Sales"].sum())
print("Average sales:", sales_df["Sales"].mean())
print("Maximum sales:", sales_df["Sales"].max())
```

Task 10: Draw a bar chart: df.plot(x="Month", y="Sales", kind="bar"). Add a title. Compare it with your Google sheets chart.

```python
import matplotlib.pyplot as plt

sales_df.plot(
    x="Month",
    y="Sales",
    kind="bar",
    legend=False
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```
